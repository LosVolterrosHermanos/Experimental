```python
import jax.numpy as jnp

import matplotlib.pyplot as plt
import numpy as np

def wsd_schedule(t, train_steps, warmup_fraction, decay_fraction):
    return np.minimum(
        np.minimum(
            t / (train_steps * warmup_fraction),
            (1.0 - (t / train_steps)) / (1.0 - decay_fraction)
        ),
        1.0
    )

# Parameters
train_steps = 10000
warmup_fraction = 0.02
decay_fraction = 0.2

# Generate t values
t_vals = np.arange(0, train_steps + 1)
wsd_vals = wsd_schedule(t_vals, train_steps, warmup_fraction, decay_fraction)

plt.figure(figsize=(8, 4))
plt.plot(t_vals, wsd_vals)
plt.title("WSD Schedule")
plt.xlabel("Step")
plt.ylabel("Schedule Value")
plt.grid(True)
plt.show()
```
